## ✍️ 1. Exercici 1: Mínim i màxim

Escriu un programa que demani un número a l'usuari repetidament fins que aquest introdueixi -99. Quan hagi acabat d'introduir els números mostra per pantalla el mínim i el màxim de la llista.

```python
def min_max() -> None:
    """
    Demana nombres enters a l'usuari i mostra el mínim i el màxim introduïts.

    La introducció de -99 indica que s'ha acabat la seqüència de nombres.

    Parameters
    ----------
    No té paràmetres.

    Returns
    -------
    None
        La funció mostra el mínim i el màxim per pantalla.
    """

    userInput: int = 0
    userList: list[int] = []
    
    while int(userInput) != -99:
        userInput = input("Introdueix un nombre enter: ")
        try:
            userInput = int(userInput)
            userList.append(userInput)
        except ValueError:
            print("L'entrada no és vàlida")
            userInput=0
        
    userList.sort()
    print(f"Mínim: {userList[0]}, Màxim: {userList[-1]}")

```

## ✍️ 2. Exercici 2: Random

Escriu un programa en el que donades una seqüència de lletres com a cadena i un número n, retorna una cadena de llargada n formada per caràcters aleatoris de la seqüència de lletres.

```python
import random


def aleatori(s: str, n: int) -> str:
    """
    Genera una cadena aleatòria a partir dels caràcters d'una cadena donada.

    Parameters
    ----------
    s : str
        Cadena que conté els caràcters disponibles.
    n : int
        Longitud de la cadena que es vol generar.

    Returns
    -------
    resultat : str
        Cadena de longitud n formada per caràcters seleccionats aleatòriament.
    """
    resultList: list[str] = []
    for i in range(n):
        resultList.append(random.choice(s))
    return "".join(resultList)
        

def aleatori2(s: str, n: int) -> str:
    """
    Genera una cadena aleatòria mitjançant una comprensió de llista.

    Parameters
    ----------
    s : str
        Cadena que conté els caràcters disponibles.
    n : int
        Longitud de la cadena que es vol generar.

    Returns
    -------
    resultat : str
        Cadena de longitud n formada per caràcters seleccionats aleatòriament.
    """

    resultat = [random.choice(s) for i in range(n)]
    return "".join(resultat)


print(aleatori("acgt", 10))
print(aleatori2("acgt", 10))
```

## ✍️ 3. Exercici 3: Nombres narcisistes

Escriu un programa que verifiqui si un nombre és narcisista. Diem que un nombre és narcisista si és la suma dels seus dígits, cadascun elevat a la potència del nombre de dígits que té.

Exemple: 153 és narcisista perquè
    $153 = 1^3 + 5^3 + 3^3 = 1 + 125 + 27$
    
Font: Adaptat de [Checking for Narcissism in Numbers / Vanessa Osuka](https://code.likeagirl.io/algorithm-part-1-checking-for-narcissism-of-a-number-python-cbea596b2790)

```python
def narcisista(n: int) -> bool:
    """
    Comprova si un nombre és narcisista.

    Un nombre és narcisista si és igual a la suma dels seus dígits
    elevats a la potència del nombre de dígits.

    Parameters
    ----------
    n : int
        Nombre enter que es vol comprovar.

    Returns
    -------
    resultat : bool
        True si el nombre és narcisista; False en cas contrari.
    """

    resultat: int = 0
    nstr = str(n)
    for letter in nstr:
        resultat += (int(letter))**(len(nstr))
    if resultat == n:
        return True
    else:
        return False

numero: int = 153

if narcisista(numero):
    print(f"{numero} és narcisista")
else:
    print(f"{numero} no és narcisista")
```

## ✍️ 4. Exercici 4: Pedra, paper, tisores

Escriu una funcio que permeti a dos jugadors jugar al Pedra-Paper-Tisores. 

```python
def jugar() -> None:
    """
    Permet a dos usuaris jugar una partida de pedra, paper i tisores.

    Parameters
    ----------
    No té paràmetres.

    Returns
    -------
    None
        La funció mostra per pantalla el resultat de la partida.
    """
    
    
    jugada1: str = input(f"usuari1, tria pedra, paper o tisores: ").strip().lower()
    jugada2: str = input(f"usuari2, tria pedra, paper o tisores: ").strip().lower()

    jugades: tuple[str,str] = (jugada1,jugada2)
    match jugades:
        case ("pedra","pedra"):
            print("Empat")
        case ("tisores","tisores"):
            print("Empat")
        case ("paper","paper"):
            print("Empat")
        case ("pedra", "tisores"):
            print("Guanya el jugador 1")
        case ("tisores", "paper"):
            print("Guanya el jugador 1")
        case ("paper", "pedra"):
            print("Guanya el jugador 1")
        case ("tisores", "pedra"):
            print("Guanya el jugador 2")
        case ("tisores", "pedra"):
            print("Guanya el jugador 2")
        case ("pedra", "paper"):
            print("Guanya el jugador 2")
        case ("paper", "tisores"):
            print("Guanya el jugador 2")
        case _:
            print("Error d'entrada")
```

## ✍️ 5. Exercici 5: Endevina el nombre

Genera un nombre aleatori entre 1 i 9 (inclosos). Demana a l'usuari que l'encerti, i digues-li si ha dit un nombre massa petit, massa gran o l'ha encertat. Segueix fins que l'usuari entri 'exit', i emmagatzema el nombre d'intents de l'usuari, que mostraràs al finalitzar.

```python
import random


def encertar() -> None:
    """
    Inicia un joc en què l'usuari ha d'endevinar un nombre entre 1 i 9.

    Després de cada intent, indica si el nombre introduït és massa petit
    o massa gran. L'usuari pot escriure 'exit' per abandonar la partida.

    Parameters
    ----------
    No té paràmetres.

    Returns
    -------
    None
        La funció mostra el resultat i el nombre d'intents per pantalla.
    """

    entrada: int = 0
    intents: int = 0
    nombre = random.randint(1,9)
    jugant = True
    
    while jugant:
        entrada = int(input("Endevina el nombre: "))
        if entrada == nombre:
            print("Has encertat")
            jugant = False
        elif entrada < nombre:
            print("El nombre és més gran")
        else:
            print("El nombre és més petit")
        intents += 1
        
    print(f"Nombre d'intents: {intents}")
```
