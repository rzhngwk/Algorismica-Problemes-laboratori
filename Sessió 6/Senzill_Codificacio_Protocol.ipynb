## ✍️ Alfabet de protocol

Un protocol intern substitueix cada lletra per una paraula clau. Utilitza el diccionari següent:

`A:Arbre, B:Bosc, C:Cim, D:Delta, E:Estel, F:Far, G:Gel, H:Hora, I:Illa, J:Jardí, K:Kiwi, L:Llac, M:Mar, N:Núvol, O:Ona, P:Pont, Q:Quars, R:Riu, S:Sol, T:Terra, U:Ull, V:Vent, W:Watt, X:Xarxa, Y:Yeti, Z:Zona`.

Donada una cadena de lletres, retorna una llista amb les paraules corresponents. Ignora majúscules/minúscules.


### 🧠 Tasca


```python
def codifica_protocol(text: str) -> list[str]:
    """
    Aquesta funció converteix cada lletra d'un text
    en una paraula clau.
    
    Parameters
    ----------
    text: str
    
    Returns
    -------
    substitucio: list
    """ 
    dic: dict[str] = {"A":"Arbre", "B":"Bosc", "C":"Cim", "D":"Delta", "E":"Estel", "F":"Far", "G":"Gel", "H":"Hora", "I":"Illa", 
                      "J":"Jardí", "K":"Kiwi", "L":"Llac", "M":"Mar", "N":"Núvol", "O":"Ona", "P":"Pont", "Q":"Quars", "R":"Riu", 
                      "S":"Sol", "T":"Terra", "U":"Ull", "V":"Vent", "W":"Watt", "X":"Xarxa", "Y":"Yeti", "Z":"Zona"}
    llista: list[str] = []
    text = text.upper()

    for lletra in text:
        llista.append(dic[lletra])
    return llista
```

### ✅ Asserts públics


```python
assert codifica_protocol('CAT') == ['Cim', 'Arbre', 'Terra']
assert codifica_protocol('sol') == ['Sol', 'Ona', 'Llac']
assert codifica_protocol('X') == ['Xarxa']
```
