## ✍️ Tram de sessions sense repetir identificador

Una cadena representa els identificadors de sessions observades consecutivament. Retorna el tram consecutiu més llarg en què cap identificador apareix dues vegades. Si hi ha empat, retorna el primer tram màxim trobat.

Per exemple, a la cadena 'ABCADEAF' el tram de sessions més llarg sense repeticions és 'BCADE'.

Assegura't que la funció s'executi correctament amb qualsevol cadena d'entrada, també amb la cadena buida.

Busca una solució lineal basada en una finestra mòbil.


### 🧠 Tasca


```python
def tram_sessions_uniques(registre: str) -> str:
    """
    Aquesta funció retorna el tram consecutiu més llarg
      sense repeticions.

    Parameters
    ----------
    registre: str 
        Registre de sessions

    Returns
    -------
    tram: str
        Tram consecutiu més llarg sense repeticions
    """
    act: list[str] = []
    may: list[str] = []
    correcte: bool = True
    actual: int = 0
    
    for inici in range(len(registre)):
        correcte = True
        actual = inici
        while (actual<len(registre)) and correcte:
            if registre[actual] not in act:
                act.append(registre[actual])
            else:
                correcte = False
                if len(act)>len(may):
                    may = act
                act = []
            actual += 1

    return "".join(may)
```

### ✅ Asserts públics


```python
assert tram_sessions_uniques('ABCADEAF') == 'BCADE'
assert tram_sessions_uniques('AAAA') == 'A'
assert tram_sessions_uniques('ABCDE') == 'ABCDE'
```

## ⭐ Repte


Crea una variant de l'anterior funció en què, a més de retornar el tram, també es retorni la seva posició inicial i la final.

El retorn té la forma `(tram, inici, fi)` on `fi` és inclusiu. 

Per exemple, per a la cadena buida retornaria `("", -1, -1)`, i per a la cadena 'ABCADEAF' retornaria  ('BCADE', 1, 5).

```python
def tram_sessions_amb_limits(registre: str) -> tuple[str, int, int]:
    """
    Aquesta funció retorna el tram consecutiu més llarg
      sense repeticions, i el seu límit inferior i superior.

    Parameters
    ----------
    registre: str 
        Registre de sessions

    Returns
    -------
    Tupla amb:
        tram: str
            Tram consecutiu més llarg sense repeticions
        inici: int
            Inici del tram
        fi: int
            Final del tram (inclusiu)
    """
    
    act: list[str] = []
    may: list[str] = []
    correcte: bool = True
    actual: int = 0

    inici: int = 0
    final: int = 0

    if registre == "":
        return ("", -1, -1)
        
    for i in range(len(registre)):
        correcte = True
        actual = i
        while (actual<len(registre)) and correcte:
            if registre[actual] not in act:
                act.append(registre[actual])
            else:
                correcte = False
                if len(act)>len(may):
                    may = act
                    inici = i
                    final = actual-1
                act = []
            actual += 1

    return ("".join(may), inici, final)
```

### ✅ Asserts públics del repte


```python
assert tram_sessions_amb_limits('ABCADEAF') == ('BCADE', 1, 5)
assert tram_sessions_amb_limits('') == ('', -1, -1)
```
