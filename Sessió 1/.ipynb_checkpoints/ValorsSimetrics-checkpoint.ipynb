## ✍️ Valors simètrics respecte d’un centre

Donada una llista d'enters i un valor `centre`, troba totes les parelles de valors diferents de posició que queden a la mateixa distància del centre, és a dir, parelles `(a, b)` que compleixen

$$a-centre = centre-b.$$

Retorna cada parella amb el valor menor primer i evita duplicats. Busca una solució d'ordre $O(n\log n)$ o millor.


### 🧠 Tasca


```python
def parelles_simetria(valors: list[int], centre: int) -> list[tuple[int, int]]:
    """Retorna parelles simètriques respecte del centre."""

    result: list[tuple[int,int]] = []
    valors.sort()
    
    for a in valors:
        for b in valors:
            if a != b and abs(a-centre)==abs(centre-b) and (b,a) not in result:
                result.append((a,b))
    return result
```

### ✅ Asserts públics


```python
assert parelles_simetria([14, 2, 9, 6, 18, 10], 10) == [(2, 18), (6, 14)]
assert parelles_simetria([1, 4, 7, 10, 13], 7) == [(1, 13), (4, 10)]
assert parelles_simetria([3, 8, 12], 7) == []
```

## ⭐ Repte


Retorna les **posicions originals** de les parelles, no els valors. Si hi ha duplicats, cada posició només pot participar una vegada.

### ✅ Asserts públics del repte


```python
assert indexs_simetria([14, 2, 9, 6, 18, 10], 10) == [(0, 3), (1, 4)]
```
